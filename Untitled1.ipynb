```python


import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from IPython.display import display

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    ConfusionMatrixDisplay,
    roc_curve,
    auc
)
from sklearn.preprocessing import label_binarize

sns.set_theme(style="whitegrid")

RANDOM_STATE = 42




FILE_PATH = "/content/Global_Pollution_Analysis.csv"

df = pd.read_csv(FILE_PATH)

print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)

print("\nFirst five rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nDataset information:")
df.info()



import re

df.columns = [
    re.sub(r"_+", "_", re.sub(r"[^a-z0-9]+", "_", str(col).lower())).strip("_")
    for col in df.columns
]

print("\nStandardized column names:")
print(df.columns.tolist())



def find_column(possible_names, required=False):
    """Find a column using common name variations."""
    for name in possible_names:
        if name in df.columns:
            return name


    for name in possible_names:
        for col in df.columns:
            if name in col or col in name:
                return col

    if required:
        raise ValueError(
            f"Required column not found. Tried: {possible_names}\n"
            f"Available columns: {df.columns.tolist()}"
        )

    return None


country_col = find_column(["country"])
year_col = find_column(["year"])

co2_col = find_column([
    "co2_emissions_in_mt",
    "co2_emissions",
    "co2"
], required=True)

waste_col = find_column([
    "industrial_waste_in_tons",
    "industrial_waste_in_ton",
    "industrial_waste",
    "waste_in_tons"
])

energy_recovered_col = find_column([
    "energy_recovered_in_gwh",
    "energy_recovered",
    "energy_recovery",
    "energy_recovery_in_gwh"
], required=True)

air_col = find_column([
    "air_pollution_index",
    "air_pollution"
], required=True)

water_col = find_column([
    "water_pollution_index",
    "water_pollution"
])

soil_col = find_column([
    "soil_pollution_index",
    "soil_pollution"
])

energy_consumption_col = find_column([
    "energy_consumption",
    "energy_consumption_in_gwh",
    "total_energy_consumption"
])

population_col = find_column([
    "population",
    "total_population"
])

print("\nIdentified columns:")
for name, value in {
    "Country": country_col,
    "Year": year_col,
    "CO2 emissions": co2_col,
    "Industrial waste": waste_col,
    "Energy recovered": energy_recovered_col,
    "Air pollution index": air_col,
    "Water pollution index": water_col,
    "Soil pollution index": soil_col,
    "Energy consumption": energy_consumption_col,
    "Population": population_col
}.items():
    print(f"{name}: {value}")



print("\nMissing values before cleaning:")
display(df.isnull().sum().sort_values(ascending=False))

print("\nDuplicate rows:", df.duplicated().sum())


df = df.drop_duplicates().copy()


numeric_candidates = [
    co2_col, waste_col, energy_recovered_col,
    air_col, water_col, soil_col,
    energy_consumption_col, population_col, year_col
]

for col in numeric_candidates:
    if col is not None:
        df[col] = pd.to_numeric(df[col], errors="coerce")


df = df.replace([np.inf, -np.inf], np.nan)


df = df.dropna(subset=[energy_recovered_col]).copy()

print("\nDataset shape after removing duplicates and missing target values:")
print(df.shape)



if year_col is not None:
    df["year_numeric"] = df[year_col]


if energy_consumption_col is not None and population_col is not None:
    df["energy_consumption_per_capita"] = (
        df[energy_consumption_col] /
        df[population_col].replace(0, np.nan)
    )
    print("\nEnergy consumption per capita calculated.")

else:
    print(
        "\nEnergy consumption per capita was not calculated because "
        "the dataset does not contain both energy consumption and population."
    )


pollution_cols = [
    col for col in [air_col, co2_col, water_col, soil_col]
    if col is not None
]

if len(pollution_cols) >= 2:
    temp_scaled = pd.DataFrame(
        StandardScaler().fit_transform(df[pollution_cols]),
        columns=pollution_cols,
        index=df.index
    )
    df["pollution_composite_score"] = temp_scaled.mean(axis=1)
    print("Pollution composite score calculated.")

print("\nDataset after feature engineering:")
display(df.head())



for col in df.columns:
    if pd.api.types.is_numeric_dtype(df[col]):
        median_value = df[col].median()
        if pd.notna(median_value):
            df[col] = df[col].fillna(median_value)
    else:
        mode_values = df[col].mode(dropna=True)
        if not mode_values.empty:
            df[col] = df[col].fillna(mode_values.iloc[0])

print("\nMissing values after EDA imputation:")
print(df.isnull().sum().sum())



print("\nDescriptive statistics for numerical variables:")
display(df.describe().T)

print("\nCO2 emissions statistics:")
display(df[co2_col].describe())

if waste_col is not None:
    print("\nIndustrial waste statistics:")
    display(df[waste_col].describe())

print("\nEnergy recovery statistics:")
display(df[energy_recovered_col].describe())




numeric_df = df.select_dtypes(include=np.number)

plt.figure(figsize=(14, 9))
sns.heatmap(
    numeric_df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Correlation Heatmap of Pollution and Energy Variables")
plt.tight_layout()
plt.show()




if country_col is not None:
    country_co2 = (
        df.groupby(country_col)[co2_col]
        .mean()
        .sort_values(ascending=False)
        .head(10)
    )

    plt.figure(figsize=(12, 6))
    country_co2.plot(kind="bar")
    plt.title("Top 10 Countries by Average CO2 Emissions")
    plt.xlabel("Country")
    plt.ylabel("Average CO2 Emissions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

    print("\nTop 10 countries by average CO2 emissions:")
    display(country_co2.reset_index(name="average_co2_emissions"))




if year_col is not None:
    yearly_co2 = df.groupby(year_col)[co2_col].mean().sort_index()

    plt.figure(figsize=(11, 5))
    yearly_co2.plot(kind="line", marker="o")
    plt.title("Average CO2 Emissions Across Years")
    plt.xlabel("Year")
    plt.ylabel("Average CO2 Emissions")
    plt.tight_layout()
    plt.show()

    print("\nYearly average CO2 emissions:")
    display(yearly_co2.reset_index(name="average_co2_emissions"))




if country_col is not None:
    top_countries = (
        df.groupby(country_col)[co2_col]
        .mean()
        .nlargest(10)
        .index
    )

    plt.figure(figsize=(14, 6))
    sns.boxplot(
        data=df[df[country_col].isin(top_countries)],
        x=country_col,
        y=co2_col
    )
    plt.title("CO2 Emissions Distribution Across Top 10 Countries")
    plt.xlabel("Country")
    plt.ylabel("CO2 Emissions")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()




if waste_col is not None:
    plt.figure(figsize=(9, 6))
    sns.scatterplot(
        data=df,
        x=waste_col,
        y=energy_recovered_col
    )
    plt.title("Industrial Waste vs Energy Recovered")
    plt.xlabel("Industrial Waste")
    plt.ylabel("Energy Recovered (GWh)")
    plt.tight_layout()
    plt.show()




if year_col is not None:
    yearly_energy = (
        df.groupby(year_col)[energy_recovered_col]
        .mean()
        .sort_index()
    )

    plt.figure(figsize=(11, 5))
    yearly_energy.plot(kind="line", marker="o")
    plt.title("Average Energy Recovery Across Years")
    plt.xlabel("Year")
    plt.ylabel("Average Energy Recovered (GWh)")
    plt.tight_layout()
    plt.show()



indices_to_scale = [
    col for col in [air_col, water_col, soil_col]
    if col is not None
]

if indices_to_scale:
    scaler_for_eda = StandardScaler()
    scaled_values = scaler_for_eda.fit_transform(df[indices_to_scale])

    for i, col in enumerate(indices_to_scale):
        df[col + "_scaled"] = scaled_values[:, i]

    print("\nScaled pollution indices added:")
    print([col + "_scaled" for col in indices_to_scale])



linear_features = [
    col for col in [
        air_col,
        co2_col,
        waste_col,
        water_col,
        soil_col,
        energy_consumption_col,
        population_col,
        "energy_consumption_per_capita",
        "year_numeric"
    ]
    if col is not None and col in df.columns
]


linear_features = list(dict.fromkeys(linear_features))

if len(linear_features) == 0:
    raise ValueError("No suitable features were found for Linear Regression.")

X_reg = df[linear_features].copy()
y_reg = df[energy_recovered_col].copy()


X_reg = X_reg.select_dtypes(include=np.number)

if X_reg.shape[1] == 0:
    raise ValueError("Linear Regression requires at least one numeric predictor.")


X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.20,
    random_state=RANDOM_STATE
)

regression_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler()),
    ("model", LinearRegression())
])

regression_pipeline.fit(X_train_reg, y_train_reg)

y_pred_reg = regression_pipeline.predict(X_test_reg)


mse = mean_squared_error(y_test_reg, y_pred_reg)
mae = mean_absolute_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

print("\n========== LINEAR REGRESSION RESULTS ==========")
print("Predictors:", X_reg.columns.tolist())
print(f"Mean Squared Error (MSE): {mse:.4f}")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"R-squared (R²): {r2:.4f}")

regression_results = pd.DataFrame({
    "Actual_Energy_Recovered": y_test_reg.values,
    "Predicted_Energy_Recovered": y_pred_reg
})

print("\nActual vs predicted energy recovery:")
display(regression_results.head(10))


plt.figure(figsize=(8, 6))
plt.scatter(y_test_reg, y_pred_reg, alpha=0.7)

min_value = min(y_test_reg.min(), y_pred_reg.min())
max_value = max(y_test_reg.max(), y_pred_reg.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--"
)
plt.title("Linear Regression: Actual vs Predicted Energy Recovery")
plt.xlabel("Actual Energy Recovered (GWh)")
plt.ylabel("Predicted Energy Recovered (GWh)")
plt.tight_layout()
plt.show()


residuals = y_test_reg - y_pred_reg

plt.figure(figsize=(8, 5))
plt.scatter(y_pred_reg, residuals, alpha=0.7)
plt.axhline(y=0, linestyle="--")
plt.title("Linear Regression Residual Plot")
plt.xlabel("Predicted Energy Recovered (GWh)")
plt.ylabel("Residual")
plt.tight_layout()
plt.show()




lower_threshold = df[air_col].quantile(1 / 3)
upper_threshold = df[air_col].quantile(2 / 3)

df["Pollution_Severity"] = pd.cut(
    df[air_col],
    bins=[-np.inf, lower_threshold, upper_threshold, np.inf],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

print("\nPollution severity thresholds:")
print(f"Low / Medium boundary: {lower_threshold:.4f}")
print(f"Medium / High boundary: {upper_threshold:.4f}")

print("\nPollution severity distribution:")
display(df["Pollution_Severity"].value_counts().to_frame("Count"))


classification_features = [
    col for col in [
        co2_col,
        waste_col,
        water_col,
        soil_col,
        energy_recovered_col,
        energy_consumption_col,
        population_col,
        "energy_consumption_per_capita",
        "year_numeric",
        country_col
    ]
    if col is not None and col in df.columns and col != air_col
]

classification_features = list(dict.fromkeys(classification_features))

if not classification_features:
    raise ValueError("No suitable features were found for Logistic Regression.")

X_clf = df[classification_features].copy()
y_clf = df["Pollution_Severity"].astype(str)


numeric_features = X_clf.select_dtypes(include=np.number).columns.tolist()
categorical_features = X_clf.select_dtypes(exclude=np.number).columns.tolist()


numeric_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])


categorical_transformer = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_transformer, numeric_features),
    ("categorical", categorical_transformer, categorical_features)
])

logistic_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", LogisticRegression(
        max_iter=2000,
        class_weight="balanced",
        random_state=RANDOM_STATE
    ))
])

X_train_clf, X_test_clf, y_train_clf, y_test_clf = train_test_split(
    X_clf,
    y_clf,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y_clf
)

logistic_pipeline.fit(X_train_clf, y_train_clf)

y_pred_clf = logistic_pipeline.predict(X_test_clf)

print("\n========== LOGISTIC REGRESSION RESULTS ==========")
print("Predictors:", classification_features)
print(f"Accuracy:  {accuracy_score(y_test_clf, y_pred_clf):.4f}")
print(f"Precision: {precision_score(y_test_clf, y_pred_clf, average='weighted', zero_division=0):.4f}")
print(f"Recall:    {recall_score(y_test_clf, y_pred_clf, average='weighted', zero_division=0):.4f}")
print(f"F1-score:  {f1_score(y_test_clf, y_pred_clf, average='weighted', zero_division=0):.4f}")

print("\nClassification report:")
print(classification_report(
    y_test_clf,
    y_pred_clf,
    labels=["Low", "Medium", "High"],
    zero_division=0
))




class_labels = ["Low", "Medium", "High"]

cm = confusion_matrix(
    y_test_clf,
    y_pred_clf,
    labels=class_labels
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=class_labels
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Logistic Regression Confusion Matrix")
plt.tight_layout()
plt.show()


try:
    y_score = logistic_pipeline.predict_proba(X_test_clf)
    trained_classes = logistic_pipeline.named_steps["model"].classes_

    y_test_bin = label_binarize(
        y_test_clf,
        classes=trained_classes
    )

    plt.figure(figsize=(9, 7))
    curves_plotted = 0

    for i, class_name in enumerate(trained_classes):
        if len(np.unique(y_test_bin[:, i])) < 2:
            print(f"ROC curve skipped for {class_name}: test set lacks both outcomes.")
            continue

        fpr, tpr, _ = roc_curve(y_test_bin[:, i], y_score[:, i])
        class_auc = auc(fpr, tpr)

        plt.plot(
            fpr,
            tpr,
            label=f"{class_name} (AUC = {class_auc:.2f})"
        )
        curves_plotted += 1

    plt.plot([0, 1], [0, 1], linestyle="--", label="Random classifier")
    plt.xlabel("False Positive Rate")
    plt.ylabel("True Positive Rate")
    plt.title("Multiclass ROC Curves: Pollution Severity")
    plt.legend()
    plt.tight_layout()

    if curves_plotted > 0:
        plt.show()
    else:
        plt.close()
        print("ROC curves could not be plotted for this test split.")

except Exception as e:
    print("ROC curve generation was not possible:", e)




performance_summary = pd.DataFrame([
    {
        "Model": "Linear Regression",
        "Task": "Predict energy recovery",
        "Metric": "R²",
        "Score": r2
    },
    {
        "Model": "Linear Regression",
        "Task": "Predict energy recovery",
        "Metric": "MSE",
        "Score": mse
    },
    {
        "Model": "Linear Regression",
        "Task": "Predict energy recovery",
        "Metric": "MAE",
        "Score": mae
    },
    {
        "Model": "Logistic Regression",
        "Task": "Classify pollution severity",
        "Metric": "Accuracy",
        "Score": accuracy_score(y_test_clf, y_pred_clf)
    },
    {
        "Model": "Logistic Regression",
        "Task": "Classify pollution severity",
        "Metric": "Weighted Precision",
        "Score": precision_score(y_test_clf, y_pred_clf, average="weighted", zero_division=0)
    },
    {
        "Model": "Logistic Regression",
        "Task": "Classify pollution severity",
        "Metric": "Weighted Recall",
        "Score": recall_score(y_test_clf, y_pred_clf, average="weighted", zero_division=0)
    },
    {
        "Model": "Logistic Regression",
        "Task": "Classify pollution severity",
        "Metric": "Weighted F1-score",
        "Score": f1_score(y_test_clf, y_pred_clf, average="weighted", zero_division=0)
    }
])

print("\n========== MODEL PERFORMANCE SUMMARY ==========")
display(performance_summary)




linear_model = regression_pipeline.named_steps["model"]

coefficients = pd.DataFrame({
    "Feature": X_train_reg.columns,
    "Coefficient": linear_model.coef_
}).sort_values(
    by="Coefficient",
    key=lambda x: x.abs(),
    ascending=False
)

print("\nLinear Regression coefficients:")
display(coefficients)

plt.figure(figsize=(10, 6))
sns.barplot(
    data=coefficients,
    x="Coefficient",
    y="Feature"
)
plt.title("Linear Regression: Feature Coefficients")
plt.tight_layout()
plt.show()



print("\n========== ACTIONABLE INSIGHTS ==========")

print(
    f"Average CO2 emissions: {df[co2_col].mean():.2f}"
)
print(
    f"Average energy recovered: {df[energy_recovered_col].mean():.2f} GWh"
)
print(
    f"Median energy recovered: {df[energy_recovered_col].median():.2f} GWh"
)

if country_col is not None:
    top_co2_country = (
        df.groupby(country_col)[co2_col]
        .mean()
        .idxmax()
    )
    print(
        f"Country with the highest average CO2 emissions in this dataset: "
        f"{top_co2_country}"
    )

if waste_col is not None:
    waste_energy_corr = df[[waste_col, energy_recovered_col]].corr().iloc[0, 1]
    print(
        f"Correlation between industrial waste and energy recovery: "
        f"{waste_energy_corr:.3f}"
    )

if len(pollution_cols) >= 2:
    print(
        "The pollution composite score is available for comparing relative "
        "pollution patterns in this dataset."
    )

print("\nPotential recommendations:")
print("1. Investigate waste-to-energy technologies where waste volumes are high.")
print("2. Prioritize pollution monitoring in countries with high recorded emissions.")
print("3. Improve industrial waste segregation and recycling infrastructure.")
print("4. Evaluate renewable energy and energy-recovery projects using local feasibility studies.")
print("5. Track yearly emissions and energy recovery to assess progress over time.")




print("\n================================================")
print("GLOBAL POLLUTION ANALYSIS COMPLETED SUCCESSFULLY")
print("================================================")
print("Dataset preprocessing, EDA, feature engineering,")
print("Linear Regression, Logistic Regression, evaluation,")
print("visualizations, and initial recommendations are complete.")
```
